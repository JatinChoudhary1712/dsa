```python
class Solution:
    def mergeArrays(self, a, b):
        i = j = 0
        ans = []

        while i < len(a) and j < len(b):
            if a[i] <= b[j]:
                ans.append(a[i])
                i += 1
            else:
                ans.append(b[j])
                j += 1

        return ans


# Local Testing
sol = Solution()

a = [1, 3, 5]
b = [2, 4, 6]

print(sol.mergeArrays(a, b))
```
